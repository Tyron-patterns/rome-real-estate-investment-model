## 2. Creation of the real listing dataset from real estate websites
The creation of this dataset required manual collection of information from real estate listing websites.

To make the process more structured and partially automated, a custom class, ***RealListing***, was created. The class allows each listing to be inserted using a consistent set of fields, reducing manual repetition and helping keep the dataset organized.

The inserted features are:
- Street = address name
- Price = integer
- Surface = integer
- N. of Rooms = integer
- N. of Bathrooms = integer
- External Area = B/T/G/N (Balcony, Terrace, Garden, None)
- External Surface = integer
- Floor = integer
- Levels = integer
- Parking = O/C/B/N (Outdoor, Covered, Box, None)
- Furnished = Y/N/P (Yes, No, Partially)
- Conditions = 1: New, 2: Good Condition, 3: Full Renovation Needed
- Construction Year = integer
- Property = available / bare ownership

To ensure consistency across the dataset, predefined formats and categories were used when inserting the data (e.g. categorical values encoded with fixed labels). This reduces variability due to manual input and makes the dataset more suitable for analysis.

```python
fiumicino = RealListing("Isola Sacra (FCO) - Via Trincea delle Frasche")
```

```python
fiumicino.characteristics(street = "Via Condino", price = 250, mq = 55, rooms = 3, bathrooms = 1, 
                        exteriors = "T", ex_surface = 0,  floor = 3, levels = 3, parking = "O",
                        furnished = "N", conditions = 1, construction_year = 2026, proprieta = "new construction")
```

## 3. Converting the object to a dataset
At this stage, the data is still stored within the custom class object. In order to perform further analysis, it needs to be converted into a pandas DataFrame.
This is done by applying the ***.dataset()*** method to the object, which returns the structured dataset in tabular form.

A quick check using ***.head()*** is then performed to verify that the conversion was successful and that the dataset has the expected structure.

```python
dt_fco = fiumicino.dataset()
dt_fco.head(5)
```
